```python
import pandas as pd
```

```python
df_filmes_brasileiros = pd.read_csv(r'..\datasets\filmes_brasileiros.csv')
df_filmes_brasileiros.head()
pd.set_option('display.max_rows', None)
```

```python
list(df_filmes_brasileiros.columns)
```

```python
df_filmes_brasileiros = df_filmes_brasileiros[[
 'titleId',
 'ordering',
 'title',
 'region',
 'language',
 'types',
 'attributes',
 'isOriginalTitle',
 'titleType',
 'primaryTitle',
 'originalTitle',
 'isAdult',
 'startYear',
 'endYear',
 'runtimeMinutes',
 'genres',
 'averageRating',
 'numVotes']]
```

```python
df_filmes_brasileiros.describe()
```

```python
df_filmes_brasileiros.head()
```

```python
#Analisando variável startYear, percebemos que existem valores nulos representados por '\N', o que impede a conversão direta para inteiro. Para resolver isso, podemos substituir esses valores por 0 ou outro valor que faça sentido para a análise, e então converter a coluna para inteiro.

df_filmes_brasileiros['startYear'] = df_filmes_brasileiros['startYear'].apply(lambda x: 0 if (x == '\\N' or x == '\\N') else int(x))

df_filmes_brasileiros[df_filmes_brasileiros.startYear == '\\N']
```

```python
df_filmes_brasileiros['startYear'] = df_filmes_brasileiros['startYear'].astype(int) 
```

```python
df_filmes_brasileiros.startYear.max()

print('Identificado possiveis filmes com data de lançamento.')
```

```python
#verificando filmes com lançamentos futuros

filmes_futuros = df_filmes_brasileiros[df_filmes_brasileiros['startYear'] > 2026]

print('Há {} filmes com data de lançamento futura.'.format(filmes_futuros.shape[0]))

```

```python
filmes_futuros.sample(3)
```

```python
print('Após amostra aleatória de 3 filmes com data de lançamento futura, identificamos que os dados estão corretos, indicando que esses filmes realmente possuem data de lançamento futura.')
print('Esses filmes serão desconsiderados.')
```

```python
df_filmes_brasileiros.drop(df_filmes_brasileiros[df_filmes_brasileiros['startYear'] > 2026].index, inplace=True)
```

```python
df_filmes_brasileiros[df_filmes_brasileiros['startYear'] > 2026]
```

```python
df_filmes_brasileiros['startYear'] = df_filmes_brasileiros['startYear'].replace(0,None)
```
