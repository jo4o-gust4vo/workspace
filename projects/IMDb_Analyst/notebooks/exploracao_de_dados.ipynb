carregando os datasets ao notebooks

```python
df_titles = pd.read_csv(r'..\datasets\title.akas.tsv', sep='\t')
df_titles_basic = pd.read_csv(r'..\datasets\title.basics.tsv', sep='\t')
df_titles_episode = pd.read_csv(r'..\datasets\title.episode.tsv', sep='\t')
df_titles_principals = pd.read_csv(r'..\datasets\title.principals.tsv', sep='\t')
df_titles_ratings = pd.read_csv(r'..\datasets\title.ratings.tsv', sep='\t')
```

informações dos respectivos dataframes

```python

df_titles.info();
df_titles.head()
```

```python
#verificando regiões mais comuns
lista_regioes = list(df_titles.region.value_counts().index)
print(lista_regioes)
```

```python
df_titles_brasil = df_titles[df_titles['region'] == 'BR']
df_titles_brasil.info()
df_titles_brasil.shape
```

```python
print('Há {} títulos na região brasileira.'.format(df_titles_brasil.shape[0]))
```

```python
df_title_brasil_basic =  df_titles_brasil.merge(df_titles_basic, how='left', left_on='titleId', right_on='tconst').reset_index()
df_title_brasil_basic.head()
```

```python
df_brasil_movies = df_title_brasil_basic.merge(df_titles_ratings, how='left', left_on='titleId', right_on='tconst')
df_brasil_movies.head()
```

```python
df_brasil_movies = df_brasil_movies[df_brasil_movies.titleType == 'movie']
df_brasil_movies.titleType.value_counts()
```

```python
#salvando dataset de filmes brasileiros em um arquivo csv
df_brasil_movies.to_csv(r'..\datasets\filmes_brasileiros.csv', index=False)
```
